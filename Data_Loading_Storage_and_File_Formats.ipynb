#Data Loading

Data Loading of importing data from an external source into a programming environment for processing and analysis

```python
#Creating a Dataframe

import pandas as pd

data={"Name": ["Tej", "Yash", "Yogesh", "Yuvraj", "Shubham"],
      "Age": [20, 21, 19, 17, 21],
      "Department": ["CSE", "CSE", "IT", "ECE", "CSE"],
      "Marks": [85,90,87,89,88]}

df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.tail())
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())



print("\nData types")
print(df.dtypes)



print("\nInformation:")
print(df.info())



print("\nStatistics")
print(df.describe())
```

```python
#coverting dataframe into csv file
df.to_csv("student_output.csv", index=False)
```

```python
iris_df= pd.read_csv("iris.csv")
print(iris_df.head())
```

```python
print("Shape:", iris_df.shape)
print("Columns:", iris_df.columns.tolist())


print("\nData types")
print(iris_df.dtypes)


print("\nMissing Values:")
print(iris_df.isnull().sum())


print("\nInformation:")
print(iris_df.info())


print("\nStatistics")
print(iris_df.describe())
```

#Save and Load Excel

```python
iris_df.to_excel("iris_output.xlsx", index=False)
```

```python
iris_excel=pd.read_excel("iris_output.xlsx")
print(iris_excel.head())
```

#Save and Load JSON

```python
#orient= organizes dataframe when converting into json
#indent= each rows becomes seperate json object
iris_df.to_json("iris_output.json", orient="records", indent=4)
```

```python
iris_json=pd.read_json("iris_output.json")
print(iris_json.head())
```

```python
with open("iris_output.json", "r") as file:
    print(file.read())
```

#Save and Load Parquet

```python
iris_df.to_parquet("iris_output.parquet", index=False )
```

```python
iris_parquet= pd.read_parquet("iris_output.parquet")
print(iris_parquet.head())
```

```python
import os
files= ["student_output.csv", "iris_output.xlsx" , "iris_output.json", "iris_output.parquet"]

for file in files:
    size= os.path.getsize(file)
    print(f"File: {file}, Size: {size} bytes")
```
