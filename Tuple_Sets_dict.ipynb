```python
#to store elements of tuple in different variables accordingly
student = ("Chhavi", 20, "B.tech", "CEA")
(name, age, course, branch) = student
(name, age, *details) = student
print(name)
print(age)
print(course)
print(branch)
print(name)
print(age)
print(details)
```

```python
cities= {"Delhi", "Agra"}
print(cities)
print(type(cities))
```

```python
cities.add("Mathura")
print(cities)

cities.update(["Lucknow", "Kanpur"])
print(cities)
```

```python
print(cities)
cities.discard("Lucknow")
print(cities)
```

```python
cities.remove("Delhi")
print(cities)
```

```python
cities.update(["Lucknow", "Kanpur"])
print(cities)
rem = cities.pop()
print(rem)
print(cities)
```

```python
cities.clear()
print(cities)
```

```python

A = {"Delhi", "Agra", "Mathura"}
B = {"Agra", "Delhi", "Lucknow"}
#1st method to print union between 2 elements
print(A|B)
```

```python
#2nd method to print union between 2 elements
print(A.union(B))
```

```python
#to get common elements
print(A &B)
```

```python
#to print difference
print(A - B)
```

```python
#symmetric difference- elements presend in A & B but not in both
print ( A ^ B)
```
