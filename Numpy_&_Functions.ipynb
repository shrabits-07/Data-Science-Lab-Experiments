#**Reshaping Array**

```python
data= np.arange(1,13)
a = data.reshape(3,4)
b = data.reshape(4,3)
c = data.reshape(2,-1)
print(f'1D Array:\n{data}')
print(f'3x4:\n{a}')
print(f'4x3:\n{b}')
print(f'2xinfer:\n{c}')
print(f'A shape\n{a.shape}')
print(f'B shape\n{b.shape}')
print(f'C shape\n{c.shape}')
```

**Write a function to create an array by taking input by user. Reshape the array into numpy array 6x2, 3x1 and 5x-1 matrix and if errors comes please explain the error**

```python
import numpy as np

def create_array():
    n = int(input("Enter number of elements: "))

    arr = []

    for i in range(n):
        value = int(input(f"Enter element {i+1}: "))
        arr.append(value)

    return np.array(arr)


arr = create_array()

print("Original Array:", arr)

# Reshape into 6 x 2
try:
    print("6 x 2 Matrix:")
    print(arr.reshape(6, 2))
except ValueError:
    print("Error: 6 x 2 needs exactly 12 elements.")

# Reshape into 3 x 1
try:
    print("3 x 1 Matrix:")
    print(arr.reshape(3, 1))
except ValueError:
    print("Error: 3 x 1 needs exactly 3 elements.")

# Reshape into 5 x -1
try:
    print("5 x -1 Matrix:")
    print(arr.reshape(5, -1))
except ValueError:
    print("Error: Number of elements must be divisible by 5.")
```

#**Broadcasting**

arithmetic operations on different shapes

```python
arr = np.array([1,2,3])
res = arr+1
print(res)
```

```python
#Broadcasting from 1D to 2D Array
a = np.array([2,4,6])
b = np.array([[1,3,5], [7,9,11]])
res = a+b
print(res)
```

#**Broadcasting in conditional operations**

```python

a = np.array([12,24,35,45,60,72])
b = np.array(["Adult", "Minor"])
res = np.where(a>18, b[0], b[1])
print(res)
```
